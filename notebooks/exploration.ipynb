# Shuttle Telemetry — Quick Exploration

Load a Parquet sample written by `streaming_job.py` and sanity-check the
windowed metrics. Run the producer + streaming job first so
`data/output/shuttle_metrics/` exists.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Windowed aggregates written by the streaming job
OUTPUT = "../data/output/shuttle_metrics"
df = pd.read_parquet(OUTPUT)
print(df.shape)
df.head()
```

```python
# Velocity vs altitude over time — should look like a stable orbit
# with small sensor noise.
fig, ax = plt.subplots(figsize=(10, 4))
df.plot(x="window", y=["avg_velocity_kms", "max_altitude_km"], ax=ax)
ax.set_title("Shuttle windowed metrics over time")
ax.set_ylabel("velocity (km/s) / altitude (km)")
plt.tight_layout()
plt.show()
```

```python
# How many windows had anomalies? Useful for tuning the thresholds.
print("Windows with anomalies:", int((df["anomaly_count"] > 0).sum()))
print("Total windows:", len(df))
df.describe()
```
